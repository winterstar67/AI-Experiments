```python
import requests
from tqdm import tqdm

def download_file(url, fname, chunk_size=1024):
    resp = requests.get(url, stream=True)
    total = int(resp.headers.get("content-length", 0))
    with open(fname, "wb") as file, tqdm(total=total, unit="iB", unit_scale=True) as bar:
        for data in resp.iter_content(chunk_size=chunk_size):
            size = file.write(data)
            bar.update(size)

download_file(
    "https://raw.githubusercontent.com/rowanz/hellaswag/master/data/hellaswag_val.jsonl",
    f"{SAVE_DIR}/hellaswag_val.jsonl"
)

download_file(
    "https://raw.githubusercontent.com/rowanz/hellaswag/master/data/hellaswag_val.jsonl",
    f"{SAVE_DIR}/hellaswag_test.jsonl"
)
```
